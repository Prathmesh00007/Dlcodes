```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten,Dense,Input

import time
```

```python
train = pd.read_csv(r"C:\Users\admin\Downloads\Dataset-20261112T113230Z-1-001\Dataset\Fashion\fashion-mnist_train.csv")
test = pd.read_csv(r"C:\Users\admin\Downloads\Dataset-20261112T113230Z-1-001\Dataset\Fashion\fashion-mnist_test.csv")
```

```python
x_train = train.drop("label", axis=1).values
y_train = train["label"].values

x_test = test.drop("label", axis=1).values
y_test = test["label"].values
```

```python
x_train = x_train/255.0
x_test = x_test/255.0
```

```python
x_train = x_train.reshape(-1,28,28)
x_test = x_test.reshape(-1,28,28)

print("Dataset Loaded Successfully....")
print("Training Dataset : ",x_train.shape)
print("Testing Dataset : ",x_test.shape)
```

```python
model1 = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128,activation='sigmoid'),
    Dense(64,activation='sigmoid'),
    Dense(10,activation='softmax')
])

model1.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

history1 = model1.fit(
    x_train,
    y_train,
    epochs = 10,
    validation_split = 0.2,
    batch_size = 32,
    verbose = 1
)

time1 = time.time() - start

test_loss,test_acc = model1.evaluate(x_test,y_test,verbose=0)

print("\nTraining Time ",round(time1,2)," seconds")
print("Training Accuracy : ",round(history1.history['accuracy'][-1],4))
print("Validation Accuracy : ",round(history1.history['val_accuracy'][-1],4))
print("Testing Accuracy : ",round(test_acc,4))
print("Testing Loss : ",round(test_loss,4))
```

```python
model2 = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128,activation='tanh'),
    Dense(64,activation='tanh'),
    Dense(10,activation='softmax')
])

model2.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

history2 = model2.fit(
    x_train,
    y_train,
    epochs = 10,
    validation_split = 0.2,
    batch_size = 32,
    verbose = 1
)

time2 = time.time() - start

test_loss,test_acc = model2.evaluate(x_test,y_test,verbose=0)

print("\nTraining Time ",round(time2,2)," seconds")
print("Training Accuracy : ",round(history2.history['accuracy'][-1],4))
print("Validation Accuracy : ",round(history2.history['val_accuracy'][-1],4))
print("Testing Accuracy : ",round(test_acc,4))
print("Testing Loss : ",round(test_loss,4))
```

```python
model3 = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128,activation='relu'),
    Dense(64,activation='relu'),
    Dense(10,activation='softmax')
])

model3.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

history3 = model3.fit(
    x_train,
    y_train,
    epochs = 10,
    validation_split = 0.2,
    batch_size = 32,
    verbose = 1
)

time3 = time.time() - start

test_loss,test_acc = model3.evaluate(x_test,y_test,verbose=0)

print("\nTraining Time ",round(time3,2)," seconds")
print("Training Accuracy : ",round(history3.history['accuracy'][-1],4))
print("Validation Accuracy : ",round(history3.history['val_accuracy'][-1],4))
print("Testing Accuracy : ",round(test_acc,4))
print("Testing Loss : ",round(test_loss,4))
```

```python
plt.figure(figsize=(8,7))
plt.plot(history1.history['accuracy'],label='Sigmoid Train')
plt.plot(history2.history['accuracy'],label='TANH Train')
plt.plot(history3.history['accuracy'],label='RELU Train')

plt.plot(history1.history['val_accuracy'],'--',label='Sigmoid Train')
plt.plot(history2.history['val_accuracy'],'--',label='TANH Train')
plt.plot(history3.history['val_accuracy'],'--',label='RELU Train')

plt.grid(True)
plt.legend()

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Accuracy Curve Graph")
plt.show()
```

```python
plt.figure(figsize=(8,7))
plt.plot(history1.history['loss'],label='Sigmoid Train')
plt.plot(history2.history['loss'],label='TANH Train')
plt.plot(history3.history['loss'],label='RELU Train')

plt.plot(history1.history['val_loss'],'--',label='Sigmoid Train')
plt.plot(history2.history['val_loss'],'--',label='TANH Train')
plt.plot(history3.history['val_loss'],'--',label='RELU Train')

plt.grid(True)
plt.legend()

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss Curve Graph")
plt.show()
```

```python
names = ['sigmoid','tanh','relu']
times = [time1,time2,time3]

plt.figure(figsize=(6,5))
plt.bar(names,times)

plt.title("Comparsions")
plt.grid(axis='y')
plt.xlabel("Activation Function")
plt.ylabel("Training Time")

plt.show()
```

```python
def get_gradients(model):

    x_sample = x_train[:100]
    y_sample = y_train[:100]

    with tf.GradientTape() as tape:

        predictions = model(x_sample, training=True)

        loss = tf.keras.losses.sparse_categorical_crossentropy(
            y_sample, predictions
        )

        loss = tf.reduce_mean(loss)

    gradients = tape.gradient(loss, model.trainable_variables)

    early_gradient = tf.reduce_mean(tf.abs(gradients[0])).numpy()
    later_gradient = tf.reduce_mean(tf.abs(gradients[2])).numpy()

    return early_gradient, later_gradient


sigmoid_early, sigmoid_later = get_gradients(model1)
tanh_early, tanh_later = get_gradients(model2)
relu_early, relu_later = get_gradients(model3)


# Print gradient values

print("\n========== GRADIENT RESULTS ==========")

print("Sigmoid Early Gradient:", sigmoid_early)
print("Sigmoid Later Gradient:", sigmoid_later)

print("Tanh Early Gradient:", tanh_early)
print("Tanh Later Gradient:", tanh_later)

print("ReLU Early Gradient:", relu_early)
print("ReLU Later Gradient:", relu_later)


# ==========================================
# 9. Plot Gradient Flow Comparison
# ==========================================

early = [sigmoid_early, tanh_early, relu_early]
later = [sigmoid_later, tanh_later, relu_later]

plt.figure(figsize=(8,5))

plt.plot(names, early, marker='o', label="Early Layer")
plt.plot(names, later, marker='o', label="Later Layer")

plt.yscale('log')
plt.xlabel("Activation Function")
plt.ylabel("Gradient Magnitude (Log Scale)")
plt.title("Gradient Magnitude Comparison")
plt.legend()
plt.grid(True)
plt.show()


# ==========================================
# 10. Check Inactive ReLU Neurons
# ==========================================

hidden_model = tf.keras.Model(
    inputs=model3.inputs,
    outputs=model3.layers[2].output
)

hidden_output = hidden_model.predict(x_train[:1000], verbose=0)

positive_count = (hidden_output > 0).sum(axis=0)

inactive_neurons = (positive_count == 0).sum()

print("\n========== RELU NEURON CHECK ==========")
print("Inactive ReLU Neurons:", inactive_neurons)
```
